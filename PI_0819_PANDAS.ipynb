```python
#Criar meu primeiro dataframe

#Forma mais fácil: construir a partir de outra estrtura de dadoos, ex: dicionário

produtos = {
    "codigo":[101, 102, 103, 104],
    "nome":["Monitor", "Celular", "Teclado", "Mouse"],
    "preco":[800.00, 5000.00, 200, 100.00],
    "estoque":[5, 4, 10, 15]
}
```

```python
import pandas as pd
```

```python
#Gerar dataframe
df = pd.DataFrame(produtos) #quando chama devolve um objeto, df armazena ele
```

```python
df #visualização
#Chave - coluna
#Valores - linhas
```

```python
#Propriedades
df.info()

#mostra os detalhes do dataframe
#mostra os dados faltantes
#tipo de dados
```

```python
df.shape #mostra a dimensão
```

```python
# Incluir nova coluna
#Total estoque ($) - baseada em preço

df['total Estoque'] = df['estoque'] * df['preco']
```

```python
df['total Estoque'].mean() #valor médio de custo da coluna
```

```python
#remoção de colunas irrelevantes
# Unnamed:0
# remove as colunas chamadas 'Unnamed: 0' e 'id'
df = df.drop(columns=['Unnamed: 0', 'id'])
```

```python
df
```

```python
#substituir dados faltantes

media_atrasos = df['Arrival Delay in Minutes'].mean()
print(f"Média: {media_atrasos}")
```

```python
#preencher os dadfos faltantes pela média
#fillna -- roda o dataframe na coluna in dicada verificando todos os valores NaN e trocando pelo valor indicado
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(media_atrasos)
```

```python
df
```

```python
df['Arrival Delay in Minutes'].mean()
```
